```python
##string slicing##
string="Helo Zara,Welcome to python programming"
print("first character string[0]")
print("last chacacter string[-1]")
print("last five character[:-5]")
print("last 11 charater[-11:]")
print("reverse string:",string[::-1])
start=string.index("python")
print("python:",string[start:start+6])

```

```python
##string methods##
strM="Python beginner tutorial"
##covert to uppercase##
print(strM.upper())

##convert to lowercase
print(strM.lower())

##Capitalize the sentence##
print(strM.capitalize())

##count occurrence of character't'
print(strM.count('t'))

##replace python with machine learing##
print(strM.replace("python","machine learing"))
```

```python
##Tubles (creation Modification and access)
##create two tuples##
t1=(10,20,30)
t2=(40,50,60)

##concatenate two tuples#
t_combine=t1+t2
print(t_combine)

##repeat the elements 3 times##
print(t_combine*3)

##Access the 3rd element ##
print(t_combine[2])

##Access the first three elements##
print(t_combine[:3])
##access the last three elements
print(t_combine[-3:])
```
